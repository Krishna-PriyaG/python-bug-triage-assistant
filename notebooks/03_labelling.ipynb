# Bug Category Taxonomy

## Objective

The BugsInPy dataset does not provide an explicit bug-type label suitable for supervised machine learning. Therefore, a bug-category taxonomy is introduced to create a new target variable, **`bug_category`**, for the multiclass classification task.

Unlike repository names or exception types, bug categories describe the **underlying nature of the defect** rather than the project in which it occurred or the runtime symptom it produced. This enables the development of a supervised machine learning model that predicts the type of bug represented by a code patch.

---

# Labeling Methodology

Each bug is assigned **one primary bug category** representing the principal defect addressed by the patch.

The assignment is based primarily on the behavior corrected by the bug fix rather than the exception raised or the repository in which the bug occurred.

The following information may be considered during labeling:

- Patch content and code modifications
- Modified files and modules
- Functions and classes involved
- Exception types (supporting evidence only)
- Test file context
- Added or removed validation and error-handling logic

When multiple issues appear within the same patch, the assigned label should represent the **root cause** of the defect rather than secondary consequences.

Ambiguous cases should be flagged for manual review rather than assigned solely from a single extracted feature.

---

# Bug Categories

## 1. Input Validation

Bugs caused by missing, incorrect, or insufficient validation of inputs, parameters, arguments, or values before processing.

**Examples**

- Missing boundary checks
- Invalid parameter acceptance
- Incorrect validation logic
- Missing null or range checks

---

## 2. Type Handling

Bugs involving incorrect assumptions about data types, casting, conversions, compatibility, or handling of unexpected object types.

**Examples**

- String versus numeric handling
- Incorrect type conversions
- Improper handling of `None`
- Type compatibility issues

---

## 3. Collections and Indexing

Bugs involving lists, dictionaries, arrays, iterators, indexes, slicing, keys, or collection manipulation.

**Examples**

- Invalid indexing
- Missing dictionary keys
- Incorrect slicing
- Iterator boundary errors

---

## 4. API and Configuration

Bugs involving incorrect API behavior, configuration handling, default parameters, option processing, or interactions between software components.

**Examples**

- Incorrect API contracts
- Invalid configuration handling
- Unsupported options
- Incorrect default behavior

---

## 5. File, I/O, and Data Processing

Bugs involving files, paths, parsing, serialization, deserialization, encoding, external data formats, or network responses.

**Examples**

- File path handling
- Parsing failures
- Serialization errors
- Encoding issues
- Data loading problems

---

## 6. Date and Time

Bugs involving timestamps, calendars, time zones, datetime conversions, durations, or date parsing.

**Examples**

- Timezone conversion
- Timestamp calculation
- Invalid datetime parsing
- Calendar logic

---

## 7. Numerical and Computational Logic

Bugs involving arithmetic, numerical computation, dimensions, overflow, precision, or mathematical algorithms.

**Examples**

- Overflow
- Precision loss
- Shape mismatches
- Incorrect numerical calculations

---

## 8. Program Logic and State Management

Bugs caused by incorrect control flow, state transitions, lifecycle management, caching, initialization, or general program logic that do not clearly belong to another category.

**Examples**

- Incorrect boolean logic
- State synchronization
- Cache invalidation
- Incorrect initialization
- Lifecycle management
- Faulty algorithm logic

---

## 9. Other / Ambiguous

Bugs for which the available evidence is insufficient to confidently assign one of the defined categories.

This category should be used sparingly and reviewed periodically to determine whether the taxonomy requires refinement.

---

# Category Precedence

When multiple categories appear applicable, the assigned category should represent the **root cause** of the defect rather than its observable symptom.

The recommended precedence is:

1. Input Validation
2. Type Handling
3. Collections and Indexing
4. API and Configuration
5. File, I/O, and Data Processing
6. Date and Time
7. Numerical and Computational Logic
8. Program Logic and State Management
9. Other / Ambiguous

---

# Exception Type vs. Bug Category

Exception types are **not** treated as bug categories.

An exception represents the **runtime symptom** of a defect, whereas the bug category represents the **underlying cause** of the defect.

For example:

| Exception | Bug Category |
|-----------|--------------|
| `ValueError` | Input Validation |
| `ValueError` | Date and Time |
| `KeyError` | Collections and Indexing |
| `TypeError` | Type Handling |

The same exception may occur across multiple bug categories, while a single bug may reference multiple exception types.

Exception types are therefore used as **predictive features** during machine learning rather than as the target variable.

---

# Intended Machine Learning Use

The manually annotated **`bug_category`** column will serve as the target variable (**y**) for a supervised multiclass classification model.

The trained classifier will predict the most likely bug category for previously unseen bug patches using the engineered structural and semantic features extracted during earlier stages of the project.

The predicted bug category will later become additional metadata within the repository-aware retrieval pipeline, where it can be combined with repository information, modified modules, semantic embeddings, and vector similarity search to retrieve the most relevant historical bug fixes before LLM-based debugging assistance.
