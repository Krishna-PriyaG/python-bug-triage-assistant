```python
from pathlib import Path
import pandas as pd
import re

DATA_DIR = Path("../data/raw/bugsinpy/projects")

all_rows = []

for project in DATA_DIR.iterdir():

    if not project.is_dir():
        continue

    print(f"Processing {project.name}...")

    bugs_dir = project / "bugs"

    if not bugs_dir.exists():
        print(f"Skipping {project.name} (no bugs folder)")
        continue

    for bug in bugs_dir.iterdir():

        if not bug.is_dir():
            continue

        # Read bug.info
        bug_info = {}

        with open(bug / "bug.info", "r") as f:
            for line in f.read().splitlines():
                key, value = line.split("=", 1)
                bug_info[key] = value.strip().strip('"')


        # Read patch
        patch = ""

        if (bug / "bug_patch.txt").exists():
            with open(bug / "bug_patch.txt", "r") as f:
                patch = f.read()

        elif (bug / "bug.patch").exists():
            with open(bug / "bug.patch", "r") as f:
                patch = f.read()

        else:
            print(f"Warning: No patch file for {project.name}/{bug.name}")


        # Store one row
        row = {
            "project": project.name,
            "bug_id": bug.name,
            **bug_info,#unpack the bug_info dictionary into the row
            "patch": patch,
        }

        all_rows.append(row)

# Create DataFrame

df = pd.DataFrame(all_rows)

# Feature Engineering

df["modified_files"] = df["patch"].str.findall(
    r"diff --git a/(.*?) b/"
)

df["num_files_changed"] = df["modified_files"].apply(len)

df["lines_added"] = df["patch"].apply(
    lambda x: sum(
        1
        for line in x.splitlines()
        if line.startswith("+") and not line.startswith("+++")
    )
)

df["lines_removed"] = df["patch"].apply(
    lambda x: sum(
        1
        for line in x.splitlines()
        if line.startswith("-") and not line.startswith("---")
    )
)

df["num_hunks"] = df["patch"].str.count(r"@@")

df["patch_length"] = df["patch"].apply(
    lambda x: len(x.splitlines())
)

df["has_comment"] = df["patch"].str.contains(
    r"#|//|/\*|\*/",
    regex=True
)

df["declared_classes_in_patch"] = df["patch"].apply(
    lambda x: re.findall(
        r"class\s+([A-Za-z_]\w*)",
        x
    )
)

df["declared_functions_in_patch"] = df["patch"].apply(
    lambda x: re.findall(
        r"def\s+([A-Za-z_]\w*)",
        x
    )
)

df["number_of_declared_classes_in_patch"] = (
    df["declared_classes_in_patch"].apply(len)
)

df["number_of_declared_functions_in_patch"] = (
    df["declared_functions_in_patch"].apply(len)
)

print(df.info())
df.head()
```

```python
#the bug.info key needs to be cleaned up, as it has some extra spaces and newlines. Let's clean it up.
from pathlib import Path
import pandas as pd
import re

DATA_DIR = Path("../data/raw/bugsinpy/projects")

all_rows = []

for project in DATA_DIR.iterdir():

    if not project.is_dir():
        continue

    print(f"Processing {project.name}...")

    bugs_dir = project / "bugs"

    if not bugs_dir.exists():
        print(f"Skipping {project.name} (no bugs folder)")
        continue

    for bug in bugs_dir.iterdir():

        if not bug.is_dir():
            continue


        # Read bug.info
        bug_info = {}

        with open(bug / "bug.info", "r") as f:
            for line in f.read().splitlines():

                if "=" not in line:
                    continue

                key, value = line.split("=", 1)

                key = key.strip()# FIX
                value = value.strip().strip('"')

                bug_info[key] = value


        # Read patch
        patch = ""

        if (bug / "bug_patch.txt").exists():
            with open(bug / "bug_patch.txt", "r") as f:
                patch = f.read()

        elif (bug / "bug.patch").exists():
            with open(bug / "bug.patch", "r") as f:
                patch = f.read()

        row = {
            "project": project.name,
            "bug_id": bug.name,
            **bug_info,
            "patch": patch,
        }

        all_rows.append(row)


# Create DataFrame

df = pd.DataFrame(all_rows)


# Feature Engineering

df["modified_files"] = df["patch"].str.findall(
    r"diff --git a/(.*?) b/"
)

df["num_files_changed"] = df["modified_files"].apply(len)

df["lines_added"] = df["patch"].apply(
    lambda x: sum(
        1
        for line in x.splitlines()
        if line.startswith("+") and not line.startswith("+++")
    )
)

df["lines_removed"] = df["patch"].apply(
    lambda x: sum(
        1
        for line in x.splitlines()
        if line.startswith("-") and not line.startswith("---")
    )
)

df["num_hunks"] = df["patch"].str.count(r"@@")

df["patch_length"] = df["patch"].apply(
    lambda x: len(x.splitlines())
)

df["has_comment"] = df["patch"].str.contains(
    r"#|//|/\*|\*/",
    regex=True
)

df["declared_classes_in_patch"] = df["patch"].apply(
    lambda x: re.findall(
        r"class\s+([A-Za-z_]\w*)",
        x
    )
)

df["declared_functions_in_patch"] = df["patch"].apply(
    lambda x: re.findall(
        r"def\s+([A-Za-z_]\w*)",
        x
    )
)

df["number_of_declared_classes_in_patch"] = (
    df["declared_classes_in_patch"].apply(len)
)

df["number_of_declared_functions_in_patch"] = (
    df["declared_functions_in_patch"].apply(len)
)

# Remove accidental whitespace from column names (extra safety)
df.columns = df.columns.str.strip()

print(df.info())
df.head()
```
